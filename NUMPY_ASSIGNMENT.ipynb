```python
import numpy as np
marks = np.array([245, 278, 219, 290, 256])
print("Original total marks:", marks)
sorted_marks = np.sort(marks)[::-1]
ranking = np.argsort(marks)[::-1]
print("Marks in descending order:", sorted_marks)
print("\nRanking:")
for rank, index in enumerate(ranking, start=1):
    print("Rank", rank, "- Student", index + 1,
          "- Marks:", marks[index])
```

```python
import numpy as np
marks = np.array([95, 82, 74, 61, 45])
print("Input marks:", marks)
grades = np.select(
    [
        marks >= 90,
        marks >= 80,
        marks >= 70,
        marks >= 60
    ],
    [
        "A",
        "B",
        "C",
        "D"
    ],
    default="F"
)
print("Grades:", grades)
```

```python
import numpy as np
marks = np.array([
    [78, 85, 90],
    [65, 72, 80],
    [88, 91, 84],
    [56, 62, 70],
    [95, 89, 92]
])
print("Marks of five students:")
print(marks)
total = np.sum(marks, axis=1)
average = np.mean(marks, axis=1)
highest = np.max(marks, axis=1)
lowest = np.min(marks, axis=1)
class_average = np.mean(marks)
print("\nTotal marks of each student:")
print(total)
print("\nAverage marks of each student:")
print(average)
print("\nHighest mark of each student:")
print(highest)
print("\nLowest mark of each student:")
print(lowest)
print("\nClass average:", class_average)
above_average = np.where(average > class_average)[0] + 1
print("\nStudents performing above class average:")
print(above_average)
```

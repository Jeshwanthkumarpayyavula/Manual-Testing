```python
def calculate(a, b, operation):
    if operation == "+":
        return a + b
    elif operation == "-":
        return a - b
    elif operation == "*":
        return a * b
    elif operation == "/":
        return a / b
    else:
        return "Invalid operation"
a = float(input())
b = float(input())
operation = input()
print(calculate(a, b, operation))
```

```python
def sum_numbers(*args):
    return sum(args)
print(sum_numbers(10, 20, 30, 40))
```

```python
def employee(**args):
    for key, value in args.items():
        print(key, ":", value)
employee(name="John", ID=101, department="IT", salary=50000)
```

```python
def remove_duplicates(lst):
    result = []
    for item in lst:
        if item not in result:
            result.append(item)
    return result
lst = [10, 20, 10, 30, 20, 40, 30]
print(remove_duplicates(lst))
```

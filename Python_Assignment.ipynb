```python
start = int(input())
end = int(input())
for num in range(start, end + 1):
    if num > 1:
        for i in range(2, int(num ** 0.5) + 1):
            if num % i == 0:
                break
        else:
            print(num, end=" ")
```

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)
n = int(input())
print(factorial(n))
```

```python
square = lambda x: x * x
numbers = list(map(int, input().split()))
print(list(map(square, numbers)))
```

```python
numbers = list(map(int, input().split()))
numbers = list(set(numbers))
numbers.sort()
print(numbers[-2])
```

```python
text = input()
frequency = {}
for char in text:
    frequency[char] = frequency.get(char, 0) + 1
print(frequency)
```

```python
import math
radius = float(input())
area = math.pi * radius * radius
print(area)
```

```python
text = input()
reverse = ""
for char in text:
    reverse = char + reverse
print(reverse)
```

```python
numbers = list(map(int, input().split()))
result = []
for num in numbers:
    if num not in result:
        result.append(num)
print(result)
```

```python
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)
n = int(input())
for i in range(n):
    print(fibonacci(i), end=" ")
```
